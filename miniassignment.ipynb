```python
# ACC102 Mini Assignment - Python Notebook
# Track 4: Interactive Data Analysis Tool
# Data Analysis Workflow

# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# 2. Load dataset
data = {
    "company_name": ["Apple", "Microsoft", "Google", "Amazon", "Meta"],
    "industry": ["Technology", "Technology", "Technology", "Retail", "Technology"],
    "sales": [383285, 211413, 303368, 514000, 134902]
}
df = pd.DataFrame(data)

# 3. Dataset information
print("=== Dataset Information ===")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
print("\nFull Dataset:")
print(df)

# 4. Data cleaning
print("\n=== Data Cleaning ===")
print("Missing values in each column:")
print(df.isnull().sum())

# 5. Descriptive statistics
print("\n=== Descriptive Statistics ===")
print(df.describe())

# 6. Filter technology companies
tech_firms = df[df["industry"] == "Technology"]
print("\n=== Technology Industry Firms ===")
print(tech_firms)

# 7. Visualization 1: Total sales bar chart
plt.figure(figsize=(10, 5))
plt.bar(df["company_name"], df["sales"])
plt.title("Total Sales by Company")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

# 8. Visualization 2: Industry proportion pie chart
industry_count = df["industry"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(industry_count, labels=industry_count.index, autopct='%1.1f%%')
plt.title("Industry Distribution")
plt.tight_layout()
plt.show()

# 9. Key insights
print("\n=== Key Insights ===")
print("1. Technology companies dominate the dataset.")
print("2. Amazon has the highest sales revenue.")
print("3. The average sales of all companies is", round(df["sales"].mean(), 2))
```
